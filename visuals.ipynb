```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
years = [1920,1980,1990,2000,2010,2020]
gdp = [345,432,500,612,690,830]
```

```python
plt.figure(figsize =(6,3))
plt.figure('GDP over the years')
plt.xlabel('year')
plt.ylabel('GDP')

plt.plot(years,gdp,
         linestyle = 'dotted',
         marker = 'h',
         markersize = 10,
         markerfacecolor = 'red',
         markeredgecolor = 'blue',)
plt.show()
```

```python
courses =['Data Science','Machine Learning','Deep Learning','Computer Vision','MySQL','Python']
students =[200,150,100,50,300,400]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3))

plt.title('Course vs Students',
          color='midnightblue',
          fontweight='bold',
          fontsize=15)

plt.xlabel('Courses', color='midnightblue')
plt.ylabel('No of Students', color='midnightblue')

plt.bar(x=courses,
        height=students,
        color='orange',
        edgecolor='black',
        width=0.5,
        alpha=0.7)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3))

plt.title('Course vs Students',
          color='midnightblue',
          fontweight='bold',
          fontsize=15)

plt.pie(students,
        labels=courses,
        autopct='%1.1f%%',
        explode=[0.1, 0, 0, 0, 0, 0],
        colors=['orange', 'skyblue', 'lightgreen', 'pink'],
        startangle=90,
        wedgeprops={'edgecolor': 'black'})

plt.show()
```
