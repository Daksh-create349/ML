```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3))
plt.plot(ass1["GrLivArea"], ass1["SalePrice"], linestyle="dotted", marker="o", markersize=5)
plt.title("Living Area vs Sale Price", color="orange")
plt.xlabel("Living Area", color="orange")
plt.ylabel("Sale Price", color="orange")
plt.show()
```

```python
quality = ass1["OverallQual"].value_counts().sort_index()

plt.figure(figsize=(6, 3))
plt.bar(quality.index, quality.values, color="red", edgecolor="black", width=0.5, alpha=0.5)
plt.title("Number of Houses by Quality", color="darkred", fontweight="bold")
plt.xlabel("Quality", color="darkred")
plt.ylabel("Number of Houses", color="darkred")
plt.show()
```

```python
print(ass1["OverallQual"].unique())
print(ass1["OverallQual"].nunique())
```

```python
classes = ass1["MSZoning"].value_counts()

plt.figure(figsize=(6, 3))
plt.title("Distribution of House Zones", color="orange", fontweight="bold")
plt.pie(classes, labels=classes.index, wedgeprops={"edgecolor": "black"})
plt.show()
```

```python
plt.boxplot(ass1["SalePrice"].dropna())
plt.show()
```

```python
sns.scatterplot(x=ass1["YearBuilt"], y=ass1["SalePrice"])
plt.show()
```

```python
sns.histplot(data=ass1, x="SalePrice", kde=True)
plt.show()
```

```python
sns.regplot(x="GrLivArea", y="SalePrice", data=ass1, color="mediumorchid")
plt.show()
```

```python
import statsmodels.formula.api as smf

model_data = ass1[["SalePrice", "GrLivArea"]].dropna()
model = smf.ols("SalePrice ~ GrLivArea", data=model_data).fit()
```
