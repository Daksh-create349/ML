## FITTING A LINEAR REGRESSION MODEL

```python
import statsmodels.formula.api as smf
model = smf.ols('AT~Waist', data=df).fit()
# Fit we are using for Training the model
```

```python
sns.regplot(x="Waist", y="AT", data=df, color="mediumorchid");
```
