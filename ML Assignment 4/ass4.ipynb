```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler

employees = pd.read_csv('employee_data.csv')
```

```python
print('\nORIGINAL EMPLOYEE DATA')
print(employees.head(10))

print('\nMissing values in each column:')
print(employees.isnull().sum())
```

```python
encoder = LabelEncoder()
employees['Department_Code'] = encoder.fit_transform(employees['Department'])

print('\nDEPARTMENT LABEL ENCODING')
print(
    employees[['Department', 'Department_Code']]
    .drop_duplicates()
    .sort_values('Department')
)
```

```python
city_columns = pd.get_dummies(
    employees['City'],
    prefix='City',
    dtype=int
)

employees = pd.concat([employees, city_columns], axis=1)

print('\nONE-HOT ENCODING OF CITY')
print(employees[['City'] + list(city_columns.columns)].head(10))
```

```python
minmax = MinMaxScaler()
cols = ['Age', 'Experience_Years', 'Monthly_Salary']
minmax_scaled = minmax.fit_transform(employees[cols])
for i, col in enumerate(cols):
    employees[col + '_MinMax'] = minmax_scaled[:, i]
print('\nMIN-MAX SCALING')
print(employees[['Age', 'Age_MinMax','Monthly_Salary', 'Monthly_Salary_MinMax']].head(10).round(3))
```

```python
standard = StandardScaler()
standard_values = standard.fit_transform(employees[cols])
for i, col in enumerate(cols):
    employees[col + '_Standard'] = standard_values[:, i]
print('\nSTANDARD SCALING')
print(employees[['Age', 'Age_Standard','Experience_Years', 'Experience_Years_Standard']].head(10).round(3))
```

```python
employees.to_csv('employee_data_encoded.csv', index=False)
print('\nCompleted. Open Employee_data_processed.csv to compare all columns')
```
