```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df = pd.read_csv("WC_AT.csv")

```

```python
print(df.head())
print(df.shape)
print(df.isnull().sum())
```

```python
sns.scatterplot(data=df, x="Waist", y="AT")
plt.title("Waist vs AT")
plt.xlabel("Waist")
plt.ylabel("AT")
plt.show()
```

```python
df.corr()
```

```python
import statsmodels.formula.api as smf

model = smf.ols('AT ~ Waist', data=df).fit()
```

```python
sns.regplot(
    x="Waist",
    y="AT",
    data=df,
    color="mediumorchid"
)

plt.show()
```

```python
model.summary()
```
